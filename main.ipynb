## 问题1 估计粉丝投票数
S表示season, W表示week, I表示player, V_{s,w,i}表示第s个赛季、第w周、第i个player的投票

E表示淘汰序列，e_{s,w} = i表示指定赛季、周的淘汰的选手数为第i个

T表示发生淘汰的 (s,w) 索引集合

A表示存活序列

假设粉丝投票数V之间互不干扰（贝叶斯重要假定）,服从正态分布：
$$
V_{s,w,i} \sim N(\hat{\theta_{s,w,i}}, \sigma^2)
$$

其先验联合密度函数为：

$$
P(V) = \prod_{s \in S}\prod_{w \in W}\prod_{i \in I} P(V_{s,w,i})
$$

在第s个season，在第w周淘汰第i位选手的概率为：
$$
P(e_{s,w} = i | A_{s,w},V_{s,w,.}) = \frac{exp(-V_{s,w,i})}{\sum_{j \in A_{s,w}} exp(-V_{s,w,j})}
$$

其似然函数为：
$$
P(E|V) = \prod_{(s,w)\in T} \frac{exp(-V_{s,w,e_{s,w}})}{\sum_{j \in A_{s,w}} exp(-V_{s,w,j})}
$$

后验分布有：
$$
P(V|E) \propto P(E|V) P(V)
$$

最大化后验分布即可得到V序列
